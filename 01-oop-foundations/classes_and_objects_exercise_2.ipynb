An online shopping platform issues gift cards. Each gift card has a unique card ID and a monetary balance.

Design a Python class called GiftCard supporting the following requirements.

Requirements
1. A gift card is created with a card ID and initial balance.
2. A customer can add funds to the card.
3. A customer can spend funds from the card.
4. The balance cannot become negative.
5. Adding or spending a non-positive amount must be rejected.
6. A customer can check the current balance.
Example interaction

```python
from typing import List
```

```python
class GiftCard:
    def __init__(self, id:str, bal:int):
        self._id = id
        if bal >= 0: 
            self._bal = bal
        else:
            raise ValueError("Cannot Initialize with negetive Balance")

    def add_funds(self, amount:int) -> None:
        if amount <= 0:
            raise ValueError("Enter a positive amount")
        else:
            self._bal += amount

    def spend(self, amount:int) -> None:
        if amount > 0:
            if amount > self._bal:
                raise ValueError("Amount to be withdrawn is greater than available balance")
            else:
                self._bal -= amount
        else:
            raise ValueError("Enter a positve amount")

    def get_balance(self) -> int:
        return self._bal  
```
