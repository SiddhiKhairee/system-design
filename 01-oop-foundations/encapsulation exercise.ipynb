### Thermostat

Encapsulation — Practical assessment

10–15 minutes

Design a Python class called Thermostat that manages the temperature setting of a smart-home device.

Requirements
1. The thermostat starts at 22°C.
2. Users can change the target temperature using set_temperature(temp).
3. The allowed temperature range is 16°C to 30°C, inclusive.
4. Invalid temperature requests must raise ValueError without modifying the current setting.
5. Users can read the temperature through get_temperature().
6. Users can switch the thermostat between ON and OFF using turn_on() and turn_off().
7. When OFF, changing the temperature is not allowed and must raise ValueError.
8. The thermostat starts ON.

```python
from typing import List
```

```python
class Thermostat:

    def __init__(self):
        self._is_on = True
        self._temp = 22

    def set_temperature(self, new_temperature) -> None:
        if self._is_on:
            if new_temperature >= 16 and new_temperature <= 30:
                self._temp = new_temperature
            else:
                raise ValueError("Temperature out of range")
        else:
            raise ValueError("Thermostat is OFF")

    def get_temperature(self) -> int:
        if self._is_on:
            current_temperature = self._temp
            return current_temperature

    def turn_on(self) -> None:
        self._is_on = True

    def turn_off(self) -> None:
        self._is_on = False
```

```python
t = Thermostat()

t.set_temperature(25)
print(t.get_temperature())  # 25

t.turn_off()
t.set_temperature(20)       # Raises ValueError
print(t.get_temperature())  # Still 25
```
