### Proposed Design

We have an abstract method called PaymentCalculation, which will have a method called calculate_pay()

Then we have 2 methods called FixedSalaryCalculation and HourlyPaymentCalculation which will implement calculate_pay

Our final class will be employee, this class will initialize name, and then will delected payment calculation to PaymentCalculation

```python
from abc import ABC, abstractmethod
# abstract class
class PaymentCalculator(ABC):

    @abstractmethod
    def calculate_pay(self):
        pass

class FixedSalaryCalculation(PaymentCalculator):
    def __init__(self, sal):
        self.sal = sal
    def calculate_pay(self):
        return self.sal

class HourlyPaymentCalculation(PaymentCalculator):
    def __init__(self, rate, hours):
        self.rate = rate
        self.hours = hours
    def calculate_pay(self):
        return self.rate*self.hours

class Employee:
    def __init__(self, name, pay:PaymentCalculator):
        self._name = name
        self._pay = pay

    def get_salary(self):
        return self._pay.calculate_pay()
```
